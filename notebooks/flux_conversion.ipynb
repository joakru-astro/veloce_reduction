```python
import numpy as np
from matplotlib import pyplot as plt
```

```python
def mag2flux(ref_mag):
    return 10**(-(ref_mag+48.590)/2.5)
```

```python
ref_wave, ref_mag, ref_step = np.loadtxt('hr7950.dat', unpack=True)
```

```python
Fnu = mag2flux(ref_mag)
```

```python
plt.plot(ref_wave, ref_mag)
```

```python
%matplotlib inline
```

```python
plt.plot(ref_wave, Fnu)
```

```python
np.savetxt('HR7950_Fnu.txt', np.array([ref_wave, Fnu]).T)
```

```python
lam, Flam = np.loadtxt('HR7950_Flam.txt', unpack=True)
```

```python
plt.plot(lam, Flam)
plt.show()
```
